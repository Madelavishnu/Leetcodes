# 1 - TWO SUM


```python
from typing import List


class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        '''
        dict = {}
        nums = [2,7,11,15] target = 9
        i = 0
        dict = {7:0 }     target - i[0](2) = 7
        now check if that value is in that list
        if it return [dict[i], i]

        '''

        dict1 = {}

        for i in range(len(nums)):
            if nums[i] in dict1:
                return [dict1[nums[i]],i]
            else:
                dict1[target-nums[i]] = i

nums = [2,7,5,6]
target = 9
obj = Solution()
obj.twoSum(nums,target)
```

# 9 - PALINDROME NUMBER

```python
class Solution1:
  def isPalindrome(self, x : int)->bool:
    res = x
    if x < 0:
      return False


    else:
      rev = 0
      while x>0:
        rem = x%10
        rev = rev * 10 + rem
        x = x//10

      if res == rev:
        return True
      else:
        return False



x = 10
obj = Solution1()
obj.isPalindrome(x)
```
