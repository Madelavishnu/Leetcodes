# 1 - TWO SUM


```python
from typing import List


class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        '''
        dict = {}
        nums = [2,7,11,15] target = 9
        i = 0
        dict = {7:0 }     target - i[0](2) = 7
        now check if that value is in that list
        if it return [dict[i], i]

        '''

        dict1 = {}

        for i in range(len(nums)):
            if nums[i] in dict1:
                return [dict1[nums[i]],i]
            else:
                dict1[target-nums[i]] = i

nums = [2,7,5,6]
target = 9
obj = Solution()
obj.twoSum(nums,target)
```

# 9 - PALINDROME NUMBER

```python
class Solution1:
  def isPalindrome(self, x : int)->bool:
    res = x
    if x < 0:
      return False


    else:
      rev = 0
      while x>0:
        rem = x%10
        rev = rev * 10 + rem
        x = x//10

      if res == rev:
        return True
      else:
        return False



x = 10
obj = Solution1()
obj.isPalindrome(x)
```

# 13 - ROMAN TO INTEGER

```python
class Solution:
    def romanToInt(self, s: str) -> int:

      '''
      * create roman number
      * iterate to end
      * check the given one is in roman number or not
      * conditions : if roman num[s[i]] > roman num [s[i]] --> add it
                    else subtract that'''
      roman_map = {

            'M' : 1000,
            'D' : 500,
            'C' : 100,
            'L' : 50,
            'X' : 10,
            'V' : 5,
            'I' : 1
             }


      if len(s) == 0:
        return -1
      else:
        sum = 0

        for i in range(len(s)):
          if s[i] in roman_map:
              if i + 1 < len(s) and roman_map[s[i]] < roman_map[s[i + 1]]:
                sum -= roman_map[s[i]]
              else:
                sum += roman_map[s[i]]
          return sum

s = "MCMXCIV"

obj2 = Solution()
obj2.romanToInt(s)
```
